## List 연습 노트 파일입니다.

리스트의 합병 연산 (+)
리스트의 곱 연산 (*)

```python
fruit1 = ["사과", "바나나"]
fruit2 = ["복숭아", "포도"]
print(fruit1)
```

```python
temps = [28, 31, 33, 35, 27, 26, 25]
values = list(temps)                    # 얕은 복사
temps[0] = 50
print(temps)
print(values)
```

```python
temps = [[1,2],3,4,5]
values = list(temps)
print(temps)
print(values)

temps[0][0] = 7          # 리스트 안의 구조화된 객체는 참조만 복사됨. (1차원 객체들은 복사하나 더 깊은 것은 참조만 복사)
print(temps)
print(values)
```
