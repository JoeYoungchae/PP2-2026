```python
import tkinter as tk
# --- 팝업창(Toplevel)을 띄우는 함수 ---
def open_setting_window():
    # 1. 새 창 생성 (root를 부모로 지정)
    popup = tk.Toplevel(root)
    popup.title("환경설정")
    popup.geometry("300x150")
    # [옵션] 메인 창의 정중앙 근처에 뜨도록 좌표 설정
    popup.geometry(f"+{root.winfo_x() + 50}+{root.winfo_y() + 50}")
    # [옵션] 팝업창이 닫히기 전까지 메인 창을 클릭하지 못하게 독점 (Modal 효과)
    popup.grab_set()
    # 2. 팝업창 내부 위젯 배치
    lbl = tk.Label(popup, text="새로운 사용자 이름을 입력하세요:")
    lbl.pack(pady=10)
    entry = tk.Entry(popup, width=20)
    entry.pack(pady=5)
    entry.insert(0, main_label["text"]) # 현재 메인창 이름을 기본값으로 세팅

# 3. 데이터 적용 및 창 닫기 함수
def apply_and_close():
    new_name = entry.get()
    main_label.config(text=new_name) # 메인 창의 라벨 변경
    popup.destroy() # 팝업 창 제거
    btn_save = tk.Button(popup, text="저장 및 닫기", command=apply_and_close)
    btn_save.pack(pady=10)
    
# --- 메인 윈도우(Root) 설정 ---
root = tk.Tk()
root.title("메인 프로그램")
root.geometry("400x200")
# 메인 창 위젯
main_label = tk.Label(root, text="기존 사용자", font=("Arial", 16, "bold"))
main_label.pack(pady=30)
# 버튼을 누르면 open_setting_window 함수가 실행되면서 Toplevel 창이 뜸
btn_open = tk.Button(root, text="이름 변경하기 (팝업창 열기)",
command=open_setting_window, bg="lightblue")
```

```python
# Lab 숫자 추측 게임

from tkinter import *
import random

answer = random.randint(1, 100)

def guessing():
    # 텍스트 필드에서 사용자가 입력한 값을 가져온다.
    guess = int(guessField.get())

    if guess > answer:
        msg = "높음!"
    elif guess < answer:
        msg = "낮음!"
    else:
        msg = "정답!"

        resultLabel['text'] = msg
        guessField.delete(0, 5)

def reset():
    global answer
    answer = random.randint(1, 100)
    resultLabel['text'] = "다시 한 번 하세요!"

window = Tk()
window.configure(bg="white")
window.title('숫자를 맞춰보세요!')

window.geometry("500x80")
titleLabel = Label(window, text='숫자 게임에 오신 걸 환영합니다', bg="white")
titleLabel.pack()

guessField = Entry(window)
guessField.pack(side='left')
tryButton = Button(window, text='시도', fg='green', bg='white', command=guessing)
tryButton.pack(side='left')

resetButton = Button(window, text='초기화', fg='red', bg='white', command=reset)
resetButton.pack(side='left')
resultLabel = Label(window, text='1부터 100 사이의 숫자를 입력하시오.', bg='white')
resultLabel.pack(side='left')

window.mainloop()
```

```python
# Lab 가위바위보 게임

impoort random
from tkinter import *

window = Tk()
Label(window, text='선택하세요', font=("Helvetica", "16")).pack()
frame = Frame(window)

rock_image = PhotoImage(file='d:\\rock.png')
paper_image = PhotoImage(file='d:\\paper.png')
scissors_image = PhotoImage(file='d:\\scissors.png')

def pass_s():
    decide("가위")
def pass_r():
    decide("바위")
def pass_p():
    decide('보')

rock = Button(frame, image=rock_image, command=pass_r)
rock.pack(side='left')
paper = Button(frame, image=paper_image, command=pass_p)
paper.pack(side='left')
scissors = Button(frame, image=scissors_image, command=pass_s)
scissors.pack(side='left')

frame.pack()
Label(window, text='컴퓨터는 다음을 선택했습니다.', font=("Helvetica", "16")).pack()

computer_image = Label(window, image=rock_image)
computer_image.pack()
output = Label(window, text="", font=("Helvetica", "16"))
output.pack()

def decide(human):
    computer = random.choice(['가위', '바위', '보'])
    if computer == '바위':
        computer_image.configure(image=rock_image)
    elif computer == '보':
        computer_image.configure(image=paper_image)
    else:
        computer_image.configure(image=scissors_image)

    if (computer == '바위' and human == '보') or (computer == '보' and human == '가위') or (computer == '가위' and human == '바위'):
        result = "인간 승리!"
    elif computer == human:
        result = "비겼습니다."
    else:
        result = "컴퓨터 승리!"
    output.config(text='인간: ' + human + ' 컴퓨터: ' + computer + "   " + result)

window.mainloop()
```

```python
# Lab TIC-TAC-TOE 게임

from tkinter import *

# i번째 버튼을 누를 수 있는지 검사. 누를 수 있으면 O나 X를 표시
def checked(i):
    global player
    button = list[i]    # 리스트에서 i번째 버튼 객체를 가져온다.

# 버튼이 초기상태가 아니면 이미 누른 버튼이므로 아무것도 하지 않고 리턴한다.
if button['text'] != "      ":
    return
button['text'] = "  " + player + "  "
button['bg'] = 'yellow'
if player == "X":
    player = "O"
    button['bg'] = 'yellow'
else:
    player = 'X'
    button['bg'] = 'lightgreen'

window = Tk()
player = 'X'
list = []

# 9개의 버튼을 생성하여 격자 형태로 윈도우에 배치
for i in range(9):
    b = Button(window, text="       ", command=lambda k=i: checked(k))
    b.grid(row=i//3, column=i%3)
    list.append(b)

window.mainloop()
```

```python
# Lab 도형 그리기

from tkinter import *

WIDTH = 600
HEIGHT = 200
def displayRect():
    canvas.create_rectangle(10, 10, WIDTH-10, HEIGHT-10)

def displayOval():
    canvas.create_oval(10, 10, WIDTH-10, HEIGHT-10, fill='yellow')

def displayArc():
    canvas.create_arc(10, 10, WIDTH-10, HEIGHT-10, start=0, extent=120, width=10, fill='blue')

def displayPolygon():
    canvas.create_polygon(10, 10, WIDTH-10, HEIGHT-10, 200,90,300,160)

def displayLine():
    canvas.create_line(10,10,WIDTH-10,HEIGHT-10,fill='green')

def clearCanvas():
    canvas.delete(ALL)

window=Tk()
canvas=Canvas(window, width=WIDTH, height=HEIGHT, bg='white')
canvas.pack()
frame=Frame(window)
frame.pack()

btRectangle=Button(frame, text="Rectangle",
command=displayRect).grid(row=1,column=2)
btOval=Button(frame,text="Oval",command=displayOval).grid(row=1,column=3)
btArc=Button(frame, text="Arc",command=displayArc).grid(row=1,column=5)
btPolygon=Button(frame,
text="Polygon",command=displayPolygon).grid(row=1,column=4)
btLine=Button(frame, text="Line",command=displayLine).grid(row=1,column=1)
btClear=Button(frame,text="Clear",command=clearCanvas).grid(row=1,column=7)

window.mainloop()
```
