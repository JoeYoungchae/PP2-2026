```python
import tkinter as tk
# --- 팝업창(Toplevel)을 띄우는 함수 ---
def open_setting_window():
    # 1. 새 창 생성 (root를 부모로 지정)
    popup = tk.Toplevel(root)
    popup.title("환경설정")
    popup.geometry("300x150")
    # [옵션] 메인 창의 정중앙 근처에 뜨도록 좌표 설정
    popup.geometry(f"+{root.winfo_x() + 50}+{root.winfo_y() + 50}")
    # [옵션] 팝업창이 닫히기 전까지 메인 창을 클릭하지 못하게 독점 (Modal 효과)
    popup.grab_set()
    # 2. 팝업창 내부 위젯 배치
    lbl = tk.Label(popup, text="새로운 사용자 이름을 입력하세요:")
    lbl.pack(pady=10)
    entry = tk.Entry(popup, width=20)
    entry.pack(pady=5)
    entry.insert(0, main_label["text"]) # 현재 메인창 이름을 기본값으로 세팅

# 3. 데이터 적용 및 창 닫기 함수
def apply_and_close():
    new_name = entry.get()
    main_label.config(text=new_name) # 메인 창의 라벨 변경
    popup.destroy() # 팝업 창 제거
    btn_save = tk.Button(popup, text="저장 및 닫기", command=apply_and_close)
    btn_save.pack(pady=10)
    
# --- 메인 윈도우(Root) 설정 ---
root = tk.Tk()
root.title("메인 프로그램")
root.geometry("400x200")
# 메인 창 위젯
main_label = tk.Label(root, text="기존 사용자", font=("Arial", 16, "bold"))
main_label.pack(pady=30)
# 버튼을 누르면 open_setting_window 함수가 실행되면서 Toplevel 창이 뜸
btn_open = tk.Button(root, text="이름 변경하기 (팝업창 열기)",
command=open_setting_window, bg="lightblue")
```

```python
# Lab 숫자 추측 게임

from tkinter import *
import random

answer = random.randint(1, 100)

def guessing():
    # 텍스트 필드에서 사용자가 입력한 값을 가져온다.
    guess = int(guessField.get())

    if guess > answer:
        msg = "높음!"
    elif guess < answer:
        msg = "낮음!"
    else:
        msg = "정답!"

        resultLabel['text'] = msg
        guessField.delete(0, 5)

def reset():
    global answer
    answer = random.randint(1, 100)
    resultLabel['text'] = "다시 한 번 하세요!"

window = Tk()
window.configure(bg="white")
window.title('숫자를 맞춰보세요!')

window.geometry("500x80")
titleLabel = Label(window, text='숫자 게임에 오신 걸 환영합니다', bg="white")
titleLabel.pack()

guessField = Entry(window)
guessField.pack(side='left')
tryButton = Button(window, text='시도', fg='green', bg='white', command=guessing)
tryButton.pack(side='left')

resetButton = Button(window, text='초기화', fg='red', bg='white', command=reset)
resetButton.pack(side='left')
resultLabel = Label(window, text='1부터 100 사이의 숫자를 입력하시오.', bg='white')
resultLabel.pack(side='left')

window.mainloop()
```

```python
# Lab TIC-TAC-TOE 게임

from tkinter import *

# i번째 버튼을 누를 수 있는지 검사. 누를 수 있으면 O나 X를 표시
def checked(i):
    global player
    button = list[i]    # 리스트에서 i번째 버튼 객체를 가져온다.

    # 버튼이 초기상태가 아니면 이미 누른 버튼이므로 아무것도 하지 않고 리턴한다.
    if button['text'] != "      ":
        return
    button['text'] = "  " + player + "  "
    button['bg'] = 'yellow'
    if player == "X":
        player = "O"
        button['bg'] = 'yellow'
    else:
        player = 'X'
        button['bg'] = 'lightgreen'

window = Tk()
player = 'X'
list = []

# 9개의 버튼을 생성하여 격자 형태로 윈도우에 배치
for i in range(9):
    b = Button(window, text="       ", command=lambda k=i: checked(k))
    b.grid(row=i//3, column=i%3)
    list.append(b)

window.mainloop()
```

```python
# 마우스 이벤트 처리

from tkinter import *

window = Tk()
window.geometry("600x200")

def callback(event):
    print(event.x, event.y, "에서 마우스 이벤트 발생")

window.bind("<Button-1>", callback)
window.mainloop()
```

#### 이벤트 지정자

지정자|설명
---|---
\<Button-1>|버튼 위젯이 눌렸을 때 발생하는 이벤트
\<B1-Motion>|마우스 버튼 1(왼쪽 버튼)이 눌린 채로 움직일 때 발생
\<ButtonRelease-1>|Button 1에서 손을 뗄 때 발생한다.
\<Double-Button-1>|마우스 버튼 1이 더블 클릭 시 발생.
\<Enter>|마우스 포인터가 위젯으로 진입했을 때 발생.
\<Leave>|마우스 포인터가 위젯 떠났을 때 발생
\<FocusIn>|키보드 포커스가 현재의 위젯으로 이동하였다.
\<FocusOut>|키보드 포커스가 현재 위젯에서 다른 위젯으로 이동하였다.
\<return>|엔터키 입력.
\<Key>|어떤 키라도 누르면 발생. 눌린 키는 이벤트 객체의 char 멤버에 저장.
a|"a"입력 시 발생.
\<Shift-Up>|시프트 키 누른 상태로 위쪽 화살표키 누르면 발생.
\<Configure>|위젯이 크기 변경했을 때 발생.

```python
from tkinter import *

window = Tk()

def key(event):
    print(repr(event.char), "가 눌렸습니다.")

def callback(event):
    frame.focus_set()
    print(event.x, event.y, "에서 마우스 이벤트 발생")

    frame = Frame(window, width=100, height=100)
    frame.bind("<Key>", key)
    frame.bind("<Button-1>", callback)
    frame.pack()

    window.mainloop()
```

```python
##
#	이 프로그램은 페인트 프로그램을 구현한다. 
#

from tkinter import *
from tkinter.colorchooser import askcolor

DEFAULT_PEN_SIZE = 1.0
DEFAULT_COLOR = "black"

mode = "pen"
old_x = None
old_y = None
mycolor = DEFAULT_COLOR
erase_on = False

def use_pen():		# 펜 버튼이 선택되면 모드를 ”pen"으로 바꾼다. 
    global mode
    mode = "pen"
    
def use_brush():	# 브러쉬 버튼이 선택되면 모드를 ”brush"으로 바꾼다. 
    global mode
    mode = "brush"

def choose_color():	# 색상 버튼이 선택되면 사용자한테 색상을 요구한다. 
    global mycolor
    mycolor = askcolor(color=mycolor)[1]

def use_eraser():	# 지우개 버튼이 선택되면 모드를 ”erase"으로 바꾼다. 
    global mode
    mode = "erase"

def paint(event):	# 이전 점과 현재 점 사이를 직선으로 연결한다. 
    global var, erase_on, mode, old_x, old_y
    fill_color = 'white' if   mode == "erase" else mycolor
    if   old_x and old_y:
        canvas.create_line(old_x, old_y, event.x, event.y,
                           capstyle=ROUND, width=var.get(), fill=fill_color)
    old_x = event.x
    old_y = event.y

def reset(event):	# 사용자가 마우스 버튼에서 손을 떼면 이전 점을 삭제한다. 
    global old_x, old_y
    old_x, old_y = None, None

def clear_all():	# 캔버스에 그려진 모든 그림을 삭제한다. 
    global cansvas
    canvas.delete('all')
    
window = Tk()	
var = DoubleVar()	# 슬라이더와 연결될 객체를 생성한다. 

penButton = Button(window, text='펜', command=use_pen)
penButton.grid(row=0, column=0, sticky=W+E)

brushButton = Button(window, text='브러쉬', command=use_brush)
brushButton.grid(row=0, column=1, sticky=W+E)

colorButton = Button(window, text='색상선택', command=choose_color)
colorButton.grid(row=0, column=2, sticky=W+E)

eraseButton = Button(window, text='지우개', command=use_eraser)
eraseButton.grid(row=0, column=3, sticky=W+E)

clearButton = Button(window, text='모두삭제', command=clear_all)
clearButton.grid(row=0, column=4, sticky=W+E)

scale = Scale(window, variable=var, orient=VERTICAL)
scale.grid(row=1, column=5, sticky=N+S)

canvas = Canvas(window, bg='white', width=600, height=400)
canvas.grid(row=1, columnspan=5)

canvas.bind('<B1-Motion>', paint)
canvas.bind('<ButtonRelease-1>', reset)

window.mainloop()
```

```python
# Lab 공 애니메이션

from tkinter import *
import time
import random

window = Tk()

canvas = Canvas(window, width=600, height=400)
canvas.pack()

class Ball():
    def __init__(self, color, size):
        self.id = canvas.create_oval(0, 0, size, size, fill=color)
        self.dx = random.randint(1, 10)
        self.dy = random.randint(1, 10)

    def move(self):
        canvas.move(self.id, self.dx, self.dy)
        x0, y0, x1, y1 = canvas.coords(self.id)
        # 원이 위쪽이나 아래쪽으로 벗어났으면 dy의 부호를 반전시킨다.
        if y1 > canvas.winfo_height() or y0 < 0:
            self.dy = -self.dy
        # 원이 왼쪽이나 오른쪽으로 벗어났으면 dx의 부호를 반전시킨다.
        if x1 > canvas.winfo_width() or x0 < 0:
            self.dx = -self.dx

ball1 = Ball("Blue", 60)
ball2 = Ball("green", 100)
ball3 = Ball("orange", 80)

while True:
    ball1.move()
    ball2.move()
    ball3.move()
    window.update()
    time.sleep(0.05)

window.mainloop()
```

```python
# 공 애니메이션 II

from tkinter import *
import time
import random

window = Tk()

canvas = Canvas(window, width=600, height=400)
canvas.pack()

class Ball():
    def __init__(self, color, size):
        self.id = canvas.create_oval(0, 0, size, size, fill=color)
        self.dx = random.randint(1, 10)
        self.dy = random.randint(1, 10)

    def move(self):
        canvas.move(self.id, self.dx, self.dy)
        x0, y0, x1, y1 = canvas.coords(self.id)
        # 원이 위쪽이나 아래쪽으로 벗어났으면 dy의 부호를 반전시킨다.
        if y1 > canvas.winfo_height() or y0 < 0:
            self.dy = -self.dy
        # 원이 왼쪽이나 오른쪽으로 벗어났으면 dx의 부호를 반전시킨다.
        if x1 > canvas.winfo_width() or x0 < 0:
            self.dx = -self.dx

colors = ['red', 'orange', 'yellow', 'green', 'blue', 'indigo', 'violet']
ballList = []
for i in range(30):
    ballList.append(Ball(random.choice(colors), 60))

while True:
    for i in range(30):
        ballList[i].move()
    window.update()
    time.sleep(0.05)

window.mainloop()
```
