## 함수 작성 연습

```python
def get_area(radius:float) -> float:    # type hint로 타입 지정
    area = 3.14*radius**2
    return area

result = get_area(3)
print("반지름 3인 원의 면적: ", result)
```

```python
# 여러 개의 값 반환

def get_input() -> tuple[int, int]:
    return 2, 3

x, y = get_input()
```

- 함수 몸체는 pass 키워드를 이용해 나중에 작성 가능
- 앞에서 정의되지 않은 함수는 호출할 수 없다.
    - 하지만 함수 내에서는 아직 정의 안 된 함수를 호출 가능

```python
def main():
    result1 = get_area(3)
    print("반지름 3인 원 면적: ", result1)

def get_area(radius):
    area = 3.14*radius**2
    return area

main()
```

또, 매개 변수를 변경한다고 해서 인수가 변경되지 않는다.

```python
def set_radius(radius):
    radius = 100
    return

r = 20
set_radius(r)
print(r)
```

- 디폴트 인수: 기본값을 갖는 함수의 매개변수.
- 키워드 인수: 인수의 이름을 명시적 지정하여 값을 매개변수로 전달하는 방법.
    - 위치 인수는 반드시 키워드 인수보다 앞에 존재한다.

```python
def greet(name, msg="Hi"):
    print(name + ', ' + msg)
greet("영채")
```

매개 변수 이름 앞에 **를 사용해 가변 길이(여러 개의) 키워드 인수를 나타낸다.

---

- *는 반복 가능 객체를 언패킹할 수 있고,<br>
**는 딕셔너리 객체를 언패킹할 수 있다.

- 함수의 인수를 풀 때 주로 사용한다.

```python
def sum(a, b, c):
    print(a + b + c)

alist = [1, 2, 3]
sum(*alist)
```

- 람다 함수는 익명함수이다.

lambda 인수 : 수식

```python
func = lambda x, y : x + y
func2 = func

result = func(2, 3)
result2 = func2(2, 3)
print(result, result2)
```

```python
myList = [1, 2, 3, 4, 5, 6, 7, 8, 9]
result = list(map(lambda x: x**2, myList))
# map -> 리스트의 모든 원소에 대해 함수를 수행한다.

print(result)
```
