### 객체 지향의 3가지 핵심요소

1. 캡슐화: 데이터와 연산을 하나의 덩어리로 묶어 다룸
    - 정보 은닉을 위한 가시성: public, protected, private
        1. 다른 클래스에서 이 클래스의 데이터와 연산을 접근하거나 접근 불가로 만듦.
        2. public: 모든 클래스에서 접근가능, protected: 자식만 접근가능, private: 접근 불가

2. 상속: 부모클래스를 자식클래스가 그대로 물려받아 추가적인 코드를 작성할 수 있음
    - 부모의  private한 것은 상속은 받으나 접근 불가
        - protected는 외부 접근 불가능하나 자식이 물려받아 사용가능

3. 다형성: 함수 이름이 같더라도 문맥에 따라 다른 함수 수행
    - A.do(), B.do()처럼 함수 이름은 같으나 클래스 다르면 다른 함수

상속 정의

~~~python
class ElectricCar(Car):     # class 자식클래스(부모클래스 또는 슈퍼클래스)
    def setBatterySize(self, size):
        pass
~~~

```python
# 일반적인 자동차를 나타내는 클래스이다.
class Car:
    def __init__(self, maker, model, color, price):
        self.maker = maker
        self.model = model
        self.color = color
        self.price = price
    
    def setMake(self, maker):
        self.maker = maker

    def getMake(self):
        return self.maker
    
    def getDesc(self):
        return self.__str__()

    def __str__(self):
        return f"Maker: {self.maker}, Model: {self.model}, Color: {self.color}, Price: {self.price}"


class ElectricCar(Car):
    def __init__(self, maker, model, color, price, bat_size):
        super().__init__(maker, model, color, price)
        self.batterySize = bat_size

    def setBatterySize(self, bat_size):
        self.batterySize = bat_size

    def getBatterySize(self):
        return self.batterySize

    def __str__(self):
        return f"{super().__str__()}, Battery : {self.batterySize}kWh"

    def getDesc(self):
        return self.__str__()


class Truck(Car):
    def __init__(self, maker, model, color, price, payLoad):
        super().__init__(maker, model, color, price)
        self.payLoad = payLoad

    def __str__(self):
        return f"{super().__str__()}, payLoad : {self.payLoad}kg"


car1 = Car("Kia", "Ray", "Blue", 2000)
print(car1.getDesc())
print(car1.model)

ecar1 = ElectricCar("Tesla", "Model S", "Red", 8000, 81)
print(ecar1)
ecar1.setBatterySize(92)
print(ecar1.getDesc())

truck1 = Truck("Ford", "F-150", "Black", 4000, 1000)
print(truck1)
print(truck1.getMake())

print(type(truck1))
print(type(ecar1))
print(type(car1))

print(isinstance(truck1, Truck))
print(isinstance(ecar1, ElectricCar))
print(isinstance(car1, Car))

print(isinstance(truck1, Car))
print(isinstance(ecar1, Car))

#print(isinstance(truck1, Vehicle))
```
