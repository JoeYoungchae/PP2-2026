### 객체 지향의 3가지 핵심요소

1. 캡슐화: 데이터와 연산을 하나의 덩어리로 묶어 다룸
    - 정보 은닉을 위한 가시성: public, protected, private
        1. 다른 클래스에서 이 클래스의 데이터와 연산을 접근하거나 접근 불가로 만듦.
        2. public: 모든 클래스에서 접근가능, protected: 자식만 접근가능, private: 접근 불가

2. 상속: 부모클래스를 자식클래스가 그대로 물려받아 추가적인 코드를 작성할 수 있음
    - 부모의  private한 것은 상속은 받으나 접근 불가
        - protected는 외부 접근 불가능하나 자식이 물려받아 사용가능

3. 다형성: 함수 이름이 같더라도 문맥에 따라 다른 함수 수행
    - A.do(), B.do()처럼 함수 이름은 같으나 클래스 다르면 다른 함수

상속 정의

~~~python
class ElectricCar(Car):     # class 자식클래스(부모클래스 또는 슈퍼클래스)
    def setBatterySize(self, size):
        pass
~~~

```python
# 일반적인 자동차를 나타내는 클래스이다.
class Car:
    def __init__(self, maker, model, color, price):
        self.maker = maker
        self.model = model
        self.color = color
        self.price = price
    
    def setMake(self, maker):
        self.maker = maker

    def getMake(self):
        return self.maker
    
    def getDesc(self):
        return self.__str__()

    def __str__(self):
        return f"Maker: {self.maker}, Model: {self.model}, Color: {self.color}, Price: {self.price}"


class ElectricCar(Car):
    def __init__(self, maker, model, color, price, bat_size):
        super().__init__(maker, model, color, price)
        self.batterySize = bat_size

    def setBatterySize(self, bat_size):
        self.batterySize = bat_size

    def getBatterySize(self):
        return self.batterySize

    def __str__(self):
        return f"{super().__str__()}, Battery : {self.batterySize}kWh"

    def getDesc(self):
        return self.__str__()


class Truck(Car):
    def __init__(self, maker, model, color, price, payLoad):
        super().__init__(maker, model, color, price)
        self.payLoad = payLoad

    def __str__(self):
        return f"{super().__str__()}, payLoad : {self.payLoad}kg"


car1 = Car("Kia", "Ray", "Blue", 2000)
print(car1.getDesc())
print(car1.model)

ecar1 = ElectricCar("Tesla", "Model S", "Red", 8000, 81)
print(ecar1)
ecar1.setBatterySize(92)
print(ecar1.getDesc())

truck1 = Truck("Ford", "F-150", "Black", 4000, 1000)
print(truck1)
print(truck1.getMake())

print(type(truck1))
print(type(ecar1))
print(type(car1))

print(isinstance(truck1, Truck))
print(isinstance(ecar1, ElectricCar))
print(isinstance(car1, Car))

print(isinstance(truck1, Car))
print(isinstance(ecar1, Car))

#print(isinstance(truck1, Vehicle))
```

# 속성 변수의 가시성: public, protected(_로 시작), private(__로 시작)

```python
class Parent(object):
    def __init__(self):
        self.__private_money = 1000
        self._protected_money = 500
        self.public_money = 100

    def __str__(self):
        return f"Parent has total {self.__private_money + self._protected_money + self.public_money} money"

class Child(Parent):
    def __init__(self):
        super().__init__()

    def __str__(self):
        return f"Child : {self.__private_money + self._protected_money + self.public_money} money"

def test():
    p = Parent()
    print(p)
    c = Child()
    print(c)

    print(p.public_money)
    print(p._protected_money)
#    print(p.__private_money)

    print(f"child: {c.public_money}")
    print(f"child: {c._protected_money}")
#    print(f"child: {c.__private_money}")

test()
```

# 다중 상속: 2개 이상의 부모클래스로부터 상속받을 때

Subclass에서 부모가 한 명 이상이므로 super()보다는 부모클래스의 명칭으로 접근.<br><br>

~~~python
class Base1:
    pass
class Base2:
    pass
class MultiDerived(Base1, Base2):
    pass
~~~

```python
class Base1:
    def __init__(self, v1 = 10):
        self.attr1 = v1

    def display1(self):
        print(f"Base1: {self.attr1}")

    def display(self):
        self.display1()

class Base2:
    def __init__(self, v2 = 20):
        self.attr2 = v2

    def display2(self):
        print(f"Base2: {self.attr2}")

    def display(self):
        self.display2()

    

class C(Base1, Base2):
    def __init__(self, v1, v2, v3):
        Base1.__init__(self, v1)
        Base2.__init__(self, v2)
        self.attr3 = v3

    def print(self):
        print(f"{self.attr1}, {self.attr2}, {self.attr3}")
        self.display1()
        self.display2()
        self.display()
        Base1.display(self)
        Base2.display(self)

def test():
    c = C(v1=1, v2=2, v3=3)
    c.print()

    c2 = C(4, 5, 6)
    c2.print()

if __name__ == "__main__":
    test()
```

### Overriding:  부모의 함수(메소드) 재정의
- "자식 클래스의 메소드가 부모 클래스의 메소드를 오버라이드(재정의)한다."
- 다형성: 함수 이름은 같지만 함수가 소속된 클래스에 따라 다른 코드가 수행되는 것.(객체지향언어만 가능)
    <br>(하나의 식별자로 다양한 타입(클래스)을 처리하는 것)
    - 메소드 오버라이드도 다형성의 일종이다.

```python
import math

class Shape:
    def __init__(self, name="영채"):
        self.name = name

    def draw(self):
        print(f"Shape: {self.name}을(를) 그립니다.")

    def get_area(self):
        print(f"Shape: {self.name}의 면적을 계산합니다.")

class Circle(Shape):
    def __init__(self, radius=1, name="원"):
        super().__init__(name)
        self.radius = radius

    def draw(self):
        print(f"Circle: {self.name}을(를) 그립니다.")

    def get_area(self) -> float:
        print(f"Circle: {self.name}의 면적을 계산합니다.")
        return math.pi * self.radius ** 2

class Rectangle(Shape):
    def __init__(self, width = 1, height = 1, name="사각형"):
        super().__init__(name)
        self.width = width
        self.height = height

    def draw(self):
        print(f"Rectangle: {self.name}을(를) 그립니다.")

    def get_area(self) -> float:
        print(f"Rectangle: {self.name}의 면적을 계산합니다.")
        return self.width * self.height

def test():
    c = Circle(10)
    c.draw()
    print(c.get_area())

    r1 = Rectangle(width = 5, height = 10)
    r1.draw()
    print(r1.get_area())

    print("---------------------------------------")
    circles = [Circle(1), Circle(2), Circle(3)]
    for c in circles:
        #c.draw()
        print(c.get_area())
    print("---------------------------------------")

    print("---------------------------------------")
    rectangles = [Rectangle(1, 1), Rectangle(2, 2), Rectangle(3, 3)]
    for r in rectangles:
        #r.draw()
        print(r.get_area())
    print("---------------------------------------")

    print("---------------------------------------")
    shapes = [Circle(1), Rectangle(1, 1), Circle(20), Rectangle(2, 3)]
    for s in shapes:
        #r.draw()
        print(s.get_area())
    print("---------------------------------------")

if __name__ == "__main__":
    test()
```
