파일 모드|모드 이름|설명
---|---|---
"r"|읽기 모드|파일의 처음부터 읽는다.
"w"|쓰기 모드|파일의 처음부터 쓴다. 파일이 없으면 생성됨.<br>파일 존재 시 기존 내용 삭제됨
"a"|추가 모드|파일의 끝에 쓴다. 파일 없으면 생성된다.
"r+"|읽기와 쓰기 모드|파일에 읽고 쓸 수 있는 모드. 모드 변경 시 seek() 호출

```python
with open("file.txt", "r", encoding="utf-8") as f:
    # for line in f:
    #     print(line.strip())

    line = f.readline().strip()
    while line != "":
        print(line)
        line = f.readline().strip()

    eline = f.readline().strip()
    print(eline)
    print(f.tell())

    f.seek(0, 0)
    eline = f.readline().strip()
    print(eline)
    print(f.tell())
```
