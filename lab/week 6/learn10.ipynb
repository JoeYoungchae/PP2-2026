```python
# 문자 출현 횟수 계산

counter = [0] * 26
infile = open("file.txt", "r", encoding="utf-8")
ch = infile.read(1)
while ch !="":
    ch = ch.upper()
    if ch >= 'A' and ch <='Z':
        # ord(ch)는 문자의 유니코드를 반환.
        # 즉, ord(ch)에 ord("A")를 빼면 현재 문자의 번호를 알 수 있다.
        i = ord(ch) - ord("A")
        counter[i] += 1
    ch = infile.read(1)
print(counter)
```

#### 문자 인코딩
- 유니코드 중 가장 많이 사용되는 인코딩은 UTF-8이다.<br>각 문자를 1개에서 4개의 바이트로 가변적으로 인코딩한다.
- infile = open("input.txt", "r", encoding="utf-8")

- 텍스트 파일 처리 시 문자 인코딩이 중요한 이유는 인코딩에 따라서 동일한 파일이라도 파일을 이루는 바이트가 달라지기 때문이다.

```python
# Lab: 행맨

import random

guesses = ''
turns = 10

infile = open("3" \
"file.txt", "r", encoding='utf-8')
lines = infile.readlines()
word = random.choice(lines)

while turns > 0:
    failed = 0
    for char in word:
        if char in guesses:
            print(char, end="")
        else:
            print("_", end="")
            failed += 1
    if failed == 0:
        print("사용자 승리")
        break

    print("")
    guess = input("단어를 추측하시오: ")
    guesses += guess
    if guess not in word:
        turns -= 1
        print("틀렸음!")
        print(str(turns)+"기회가 남았음")
        if turns == 0:
            print("사용자 패배 정답음 "+word)

infile.close()
```

```python
# Lab: 파일 암호화

key = 'abcdefghijklmnopqrstuvwxyz'

# 평문을 받아서 암호화하고 암호문을 반환한다.
def encrypt(n, plaintext):
    result = ''

    for l in plaintext.lower():
        try:
            i = (key.index(l) + n) % 26
            result += key[i]
        except ValueError:
            result += l

    return result.lower()

# 암호문을 받아서 복호화하고 평문을 반환한다.
def decrypt(n, ciphertext):
    result = ''

    for l in ciphertext:
        try:
            i = (key.index(l) - n) % 26
            result += key[i]
        except ValueError:
            result += l

    return result

n = 3
text = 'The language of truth is simple.'
encrypted = encrypt(n, text)
decrypted = decrypt(n, encrypted)

print("평문: ", text)
print('암호문: ', encrypted)
print('복호문: ', decrypted)
```

#### 순차 접근과 임의 접근

- 순차 접근: 데이터를 파일의 처음부터 순차적으로 읽거나 기록함.
    - 한 번 읽은 데이터를 다시 읽으려면 현재 파일을 닫고 다시 열어야 한다는 단점이 있다.<br>
    또, 앞부분을 읽지 않고 중간이나 마지막으로 건너뛸 수도 없다.

- 임의 접근: 파일의 어느 위치에서든 읽기와 쓰기가 가능하다.

임의접근의 원리

- 파일 포인터는 읽기와 쓰기 동작이 현재 어떤 위치에서 이루어지는 지를 나타낸다.
    - 읽고 쓸 때마다 읽은 바이트 수만큼 파일 포인터는 증가
    - 다 읽었을 때는 EOF를 가리킴

- 임의로 파일포인터 위치를 바꿀 수 있다.
    - file.seek(오프셋, 기준위치) -> 기준위치로부터 offset 만큼 이동
    - file.tell() -> 현재 위치 반환

```python
# 텍스트 파일에서 몇 개의 문자를 읽은 후에 seek()를 이용해 다시 파일의 처음으로 돌아가 보자.

infile = open("file.txt", "r+", encoding="utf-8")
str = infile.read(10)   # 파일포인터 -> 10
print("읽은 문자열: ", str)
position = infile.tell()    # 현재의 파일포인터 -> 10
print("현재 위치: ", position)

position = infile.seek(0, 0)    # 파일포인터 -> 0
str = infile.read(10)
print("읽은 문자열: ", str)
infile.close()
```

```python
# 이미지 파일 복사하기
# 이미지 파일은 이진 파일이다. 즉 파일에 이진수 형식으로 데이터가 저장되어 있다.

infile = open("C:/Users/user/Pictures/the_doors.jpg", "rb")
outfile = open("kkk.png", "wb")

# 입력 파일에서 1024 바이트씩 읽어서 출력 파일에 쓴다.
while True:
    copy_buffer = infile.read(1024)
    if not copy_buffer:         # 파일의 끝이면
        break
    outfile.write(copy_buffer)

infile.close()
outfile.close()
print("the_doors.jpg를 kkk.png로 복사하였습니다.")
```

#### 객체 파일 저장 및 복원
- pickle 모듈의 dump()와 load() 메소드를 사용하면 객체데이터를 쓰고 읽을 수 있다.
- 리스트, 튜플, 딕셔너리, 셋 등 파이썬의 기본 데이터 타입 가능
- 이진수로 저장
- 다시 거꾸로 복원 가능

```python
import pickle

gameOption = {          # 파일을 닫는다.
    "Sound": 8,
    "VideoQuality": "High",
    "Money": 100000,
    "WeaponList": ["gun", "missile", "knife"]
}

file = open("save.p", "wb")     # 이진 파일 오픈
pickle.dump(gameOption, file)       # 딕셔너리를 피클 파일에 저장
file.close()                        # 파일을 닫는다.
```

```python
# pickle 객체의 load() 함수를 호출하여 이진 파일에 저장된 딕셔너리를 복원할 수 있다.

import pickle

file = open("save.p", "rb")     # 이진 파일 오픈
obj = pickle.load(file)         # 피클 파일에 딕셔너리를 로딩
print(obj)
```

#### 정규식
- 정규식이란 특정한 규칙을 가지고 있는 문자열들을 메타 문자를 이용해 표현하는 수식.
- re 모듈을 사용한다.

<br>

식|기능|설명
---|---|---
^|시작|문자열의 시작을 표시
$|끝|문자열 끝을 표시
.|문자|한 개의 문자와 일치 <br> (어떤 문자가 와도 상관없다.)
\d|숫자|한 개의 숫자와 일치
\w|문자와 숫자|한 개의 문자나 숫자와 일치
\s|공백문자|공백, 탭, 줄바꿈, 캐리지리턴 문자와 일치
\S|공백문자제외|공백 문자를 제외한 모든 문자
*|반복|0번 이상 반복<br>(몇 번 반복되어도 상관없다.)
+|반복|1번 이상 반복
[abc]|문자 범위|[abc]는 a 또는 b 또는 c를 나타낸다.
[^abc]|문자 범위|[^abc]는 a, b, c가 아닌 어떤 문자

```python
# [수강번호][수강코드][과목이름] 형식의 텍스트에서 코스 번호만을 추출해보자.

text="""101 COM PythonProgramming
102 MAT LinearAlgebra
103 ENG ComputerEnglish"""

import re
# re모듈의 rindall() 함수를 사용하면 해당 text에서 매칭되는 값들을 리스트로 돌려준다.
s = re.findall("\d+", text)
print(s)
```

#### 예외 처리
- 파이썬 실행 도중 발생하는 오류를 예외라고 부른다.
- 오류가 발생 시에 오류를 사용자에게 알려주고 모든 데이터를 저장하게 한 후에
<br>사용자가 우아하게 프로그램을 종료할 수 있도록 하는 것이 바람직하다.
- 역추적 메시지: 어디서 오류가 발생되었는지 보고하는 오류 메시지.

- 오류의 종류<br>
    1. 사용자 입력 오류: 사용자가 정수를 입력해야 하는데 실수 입력
    2. 장치 오류: 네트워크가 안 된다거나 하드 디스크 작동 실패
    3. 코드 오류: 잘못된 인덱스를 사용하여 배열에 접근

- IOError: 파일을 열 수 없으면 발생
- importError: 파이썬이 모듈을 찾을 수 없으면 발생
- ValueError: 연산이나 내장 함수에서 인수가 적절치 않은 값을 가지고 오면 발생
- KeyboardInterrupt: 사용자가 인터럽트 키를 누르면 발생(Cntr+C나 Delete)
- EOFError: 내장 함수가 파일의 끝을 만나면 발생

```python
# 0으로 나누는 예외 처리 코드

(x, y) = (2, 0)
try:
    z = x/y
except ZeroDivisionError as e:      # e는 시스템이 내보내는 예외 메시지를 출력한다.
    print(e)
```
