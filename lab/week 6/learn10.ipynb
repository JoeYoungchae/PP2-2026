#### 텍스트 파일 열고 닫기

- 작업폴더: python 파일 실행하는 위치

~~~python
infile = open("input.txt", "r")     # 파일객체 = open(파일이름, 파일모드)

~~~
---
<br>

파일 모드|모드 이름|설명
---|---|---
"r"|읽기 모드|파일의 처음부터 읽는다.
"w"|쓰기 모드|파일의 처음부터 쓴다. 파일이 없으면 생성됨.<br>파일 존재 시 기존 내용 삭제됨
"a"|추가 모드|파일의 끝에 쓴다. 파일 없으면 생성된다.
"r+"|읽기와 쓰기 모드|파일에 읽고 쓸 수 있는 모드. 모드 변경 시 seek() 호출

#### 파일에서 읽기
- 파일이 클 수도 있어서 줄 단위로 끊어 읽는다.
- readline() 메소드를 호출해 한 줄씩 읽는다.
    - (현재위치부터 줄의 끝에 도달할 때 까지 텍스트 읽어 반환함.)
    - 파일 끝에선 더 이상 데이터가 없기 때문에 공백 문자열 ""이 반환됨.

---
~~~python
infile = open("input.txt", "r", encoding="utf-8")
line = infile.readline()
while line != "":
    print(line)
    line = infile.readline()
~~~

~~~python
outfile = open("input.txt", "w")
~~~

```python
with open("file.txt", "r", encoding="utf-8") as f:
    # for line in f:
    #     print(line.strip())

    line = f.readline().strip()
    while line != "":
        print(line)
        line = f.readline().strip()

    eline = f.readline().strip()
    print(eline)
    print(f.tell())

    f.seek(0, 0)
    eline = f.readline().strip()
    print(eline)
    print(f.tell())
```

#### 다양한 텍스트 입출력 방법

- str.strip(): 문자열 좌우의 문자를 지운다.
- str.split(<구문문자>): 한 줄을 단어로 분리한다.(<구문문자>를 안 줄 경우에 공백문자)

```python
# 파일 전체 한꺼번에 읽어 리스트에 담기
# **주의: 파일이 거대하면 메모리 부족으로 프로그램이 다운될 수 있다.

infile =  open("file.txt", "r", encoding="utf-8")
s = infile.read()
print(s)
infile.close()
```

```python
# 텍스트 파일에서 문자 단위로 읽기

infile = open("file.txt", "r", encoding="utf-8")
ch = infile.read(1)
while ch != "":
    print(ch)
    ch = infile.read(1)
infile.close()
```

```python
# 문자 출현 횟수 계산

counter = [0] * 26
infile = open("file.txt", "r", encoding="utf-8")
ch = infile.read(1)
while ch !="":
    ch = ch.upper()
    if ch >= 'A' and ch <='Z':
        # ord(ch)는 문자의 유니코드를 반환.
        # 즉, ord(ch)에 ord("A")를 빼면 현재 문자의 번호를 알 수 있다.
        i = ord(ch) - ord("A")
        counter[i] += 1
    ch = infile.read(1)
print(counter)
```
