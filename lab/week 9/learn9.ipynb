# GUI 프로그래밍

- tkinter는 파이썬에서 그래픽 사용자 인터페이스(GUI)를 개발할 때 필요한 모듈이다.<br><br>
- 단순 위젯: Button, Canvas, Checkbutton, Entry, Label, Message 등이 여기 속한다.
- 컨테이너 컴포넌트: 다른 컴포넌트를 안에 포함할 수 있는 컴포넌트로서 Frame, Toplevel, LabelFrame 등이 여기 속한다.

```python
# 레이블이 있는 윈도우를 생성해보자

from tkinter import *       # tkinter 모듈 포함

window = Tk()               # 루트 윈도우 생성
label = Label(window, text="Hello tkinter")     # 레이블 위젯 생성
label.pack()                # pack()은 레이블 위젯을 윈도우에 배치
# mainloop()는 이벤트 처리 루프로서 사용자로부터 오는 마우스나 키보드 이벤트를 처리함.
window.mainloop()           # 윈도우가 사용자 동작을 대기
```

1. Frame(프레임) tk.Frame: 가장 기본적이고 널리 쓰이는 투명한 컨테이너
2. LabelFrame(라벨 프레임) tk.LabelFrame
3. Toplevel(탑레벨) tk.Toplevel: 메인 윈도우와 별개로 완전히 독립된 새로운 보조 창을 띄우는 컨테이너
4. PanedWindow(페인드 윈도우) tk.PanedWindow
5. Notebook(노트북 / 탭) tk.Notebook

#### 배치 관리자
- 압축 배치 관리자: 위젯을 화면에 배치할 때 구체적 좌표 지정 X, 물건을 상자 안에 차곡차곡 압축하여 밀어 넣듯 배치하는 방식
- 격자 배치 관리자: 위젯을 테이블 형태로 배치한다.
- 절대 배치 관리자: 위젯을 지정한 절대 좌표에 배치한다.
