# GUI 프로그래밍

- tkinter는 파이썬에서 그래픽 사용자 인터페이스(GUI)를 개발할 때 필요한 모듈이다.<br><br>
    - tkinter를 이용하면 윈도우를 생성하고 버튼이나 레이블같은 위젯을 이용하여 사용자와 상호작용하는 프로그램을 작성할 수 있다.
- 단순 위젯: Button, Canvas, Checkbutton, Entry, Label, Message 등이 여기 속한다.
- 컨테이너 컴포넌트: 다른 컴포넌트를 안에 포함할 수 있는 컴포넌트로서 Frame, Toplevel, LabelFrame 등이 여기 속한다.

```python
# 레이블이 있는 윈도우를 생성해보자

from tkinter import *       # tkinter 모듈 포함

window = Tk()               # 루트 윈도우 생성
label = Label(window, text="Hello tkinter")     # 레이블 위젯 생성
label.pack()                # pack()은 레이블 위젯을 윈도우에 배치
# mainloop()는 이벤트 처리 루프로서 사용자로부터 오는 마우스나 키보드 이벤트를 처리함.
window.mainloop()           # 윈도우가 사용자 동작을 대기
```

**from tkinter import \***
<br>tkinter 모듈은 Tk 툴킷을 사용하는데 필요한 모든 클래스와 함수, 상수를 가지고 있다.<br>* 기호는 모듈에서 모든 것을 포함하라는 의미.<br>

**window = Tk()**
<br>제일 먼저 루트 윈도우를 생성해야 한다. Tk() 호출 시 화면에 하나의 윈도우가 생성된다.<br>각 프로그램은 오직 하나의 루트 윈도우를 가져야 한다.
<br>루트 윈도우는 다른 위젯보다 먼저 생성되어야 한다.<br>

**label = Label(window, text="Hello tkinter")**
<br>레이블 위젯(Label 클래스로 나타냄.)을 생성, Label()의 첫 번째 매개변수는 항상 부모 컨테이너다.<br>(컨테이너: 위젯을 내부에 가질 수 있는 객체)
<br>이 문장이 실행되면 윈도우에 레이블이 추가된다.<br>

**label.pack()**
<br>레이블의 pack() 메소드 호출됨. 레이블을 컨테이너에 위치시킨다.
<br>레이블을 생성해도 pack() 함수를 호출하지 않으면 윈도우에 레이블이 나타나지 않는다.<br>

**window.mainloop()**
<br>mainloop()는 이벤트 처리 루프로, 사용자로부터 오는 마우스나 키보드 이벤트를 처리한다.
<br>tkinter를 사용하는 프로그램은 반드시 맨 끝에 이 문장을 가져야 한다.<br>

#### 기본 위젯들

위젯|설명
---|---
Label|텍스트를 표시하는 데 사용되는 위젯
Button|버튼을 제공하는 위젯
Entry|한 줄의 텍스트를 입력받는 위젯
Text|여러 줄로 된 텍스트를 입력받는 위젯
Frame|컨테이너 위젯
Checkbutton|2가지의 구별되는 값을 가지는 변수 표현
Listbox|선택 사항 표시
Menu|메뉴 표시. 풀다운 메뉴나 팝업 메뉴
Menubutton|메뉴 버튼. 풀다운 메뉴 가능
Message|텍스트 표시.
Radiobutton|여러 값을 가질 수 있는 변수 표시
Scale|슬라이더를 끌어서 수치 입력
Scrollbar|캔버스, 엔트리, 리스트박스, 텍스트위젯을 위한 스크롤 바
Toplevel|최상위 윈도우로 표시되는 독립적인 컨테이너 위젯
LabelFrame|경계선과 제목 가지는 프레임 위젯
PanedWindow|자식 위젯들을 크기조절 가능한 패널로 관리
Spinbox|특정 범위에서 값 선택

- 버튼 위젯: 클릭 가능한 버튼을 표시하는 데 사용된다.
<br>fg, bg 매개 변수를 사용해 색상을 변경할 수 있다.
<br>width, height 매개 변수를 사용해 버튼이나 레이블의 너비, 높이를 제어한다.

- 엔트리 위젯: 사용자가 키보드로 입력한 내용을 우리에게 전달한다.
<br>get()을 사용하면 사용자의 입력을 가져올 수 있다.
<br>delete()를 사용하여 사용자의 입력을 삭제한다.
<br>insert()를 사용하여 중간에 텍스트를 삽입한다.

#### 배치 관리자
- 배치 관리자: 컨테이너 안에 존재하는 위젯의 크기와 위치를 자동적으로 관리하는 객체.
    - 압축 배치 관리자(pack()): 위젯을 화면에 배치할 때 구체적 좌표 지정 X, 위젯을 블록으로 간주하여 컨테이너 안에 상하로 배치하는 방식
    - 격자 배치 관리자(grid()): 위젯을 테이블 형태(격자)로 배치한다.
    - 절대 배치 관리자(place()): 위젯을 지정한 절대 좌표에 배치한다.

```python
# 압축 배치 관리자를 사용한 예제

from tkinter import *
window = Tk()

Button(window, text="box #1", bg="red", fg="white").pack()
Button(window, text="box #2", bg="green", fg="black").pack()
Button(window, text="box #3", bg="orange", fg="white").pack()

window.mainloop()
```

```python
# 격자 배치 관리자를 사용한 예제

from tkinter import *
window = Tk()

b1 = Button(window, text="box #1", bg="red", fg="white")
b2 = Button(window, text="box #2", bg="green", fg="black")
b3 = Button(window, text="box #3", bg="orange", fg="white")
b4 = Button(window, text="box #4", bg="pink", fg="white")

b1.grid(row=0, column=0)
b2.grid(row=0, column=1)
b3.grid(row=1, column=0)
b4.grid(row=1, column=1)

window.mainloop()
```

```python
# 절대 위치 배치 관리자를 사용한 예제

from tkinter import *
window = Tk()

b1 = Button(window, text="box #1", bg="red", fg="white")
b1.place(x=0, y=0)
b2 = Button(window, text="box #2", bg="green", fg="black")
b2.place(x=20, y=30)
b3 = Button(window, text="box #3", bg="orange", fg="white")
b3.place(x=40, y=60)

window.mainloop()
```

```python
# 여러 배치 관리자 혼용하기

from tkinter import *

window = Tk()
window.geometry("600x100")      # 루트 윈도우 크기 지정(가로x세로)
# 윈도우 안에 프레임을 만든다.
f = Frame(window)

# 프레임 안에 버튼을 만든다.
b1 = Button(f, text="Box #1", bg="red", fg="white", width=10, height=1)
b2 = Button(f, text="box #2", bg="green", fg="black", width=10, height=1)
b3 = Button(f, text="box #3", bg="orange", fg="white", width=10, height=1)
b1.pack(side=LEFT)
b2.pack(side=LEFT)
b3.pack(side=LEFT)

l = Label(window, text="이 레이블은 버튼들 위에 배치된다.")
# 레이블, 프레임을 루트 윈도우에 압축 배치 관리자로 배치한다.
l.pack()
f.pack()

window.mainloop()
```

#### 버튼 이벤트 처리
- 버튼이 클릭되면 내부적으로 이벤트가 발생한다.
- 이벤트 발생 시 자동적으로 호출되는 함수를 지정해 놓으면 버튼이 클릭되었을 때 어떤 작업을 할 수 있다.
    - 콜백함수(또는 핸들러): 이벤트 발생 시 호출되는 함수.<br>버튼의 생성자를 호출할 때 command 매개변수에 이벤트 처리하는 함수의 이름을 지정하면 된다.
- 이벤트 구동 프로그래밍: 이벤트가 프로그램의 실행 순서를 결정하는 프로그래밍

```python
from tkinter import *

def process():
    print("버튼이 클릭되었습니다.")

window = Tk()
button = Button(window, text="클릭하세요.", command=process)
button.pack()
window.mainloop()
```

#### 화면에 그림 그리기
- 캔버스 위젯을 윈도우 위에 생성한 후에 그림을 그릴 수 있다.

```python
from tkinter import *

window = Tk()

w = Canvas(window, width=300, height=200)
w.pack()

i = w.create_rectangle(50, 25, 200, 100, , outline="blue", fill="#FA88AB")

w.coords(i, 0, 0, 100, 100)     # 좌표 변경
w.itemconfig(i, fill="blue")    # 색상을 변경한다.

#w.delete(i)        # 삭제
#w.delete(ALL)      # 모든 항목을 삭제
window.mainloop()
```

```python
import tkinter as tk
from tkinter import ttk, messagebox


def on_button_click(name):
    messagebox.showinfo("알림", f"'{name}' 버튼이 눌렸습니다.")


# Main window
root = tk.Tk()
root.title("Widgets Test")
root.geometry("620x720")
root.configure(bg="#CECAD3")

# Menu bar
menubar = tk.Menu(root)
filemenu = tk.Menu(menubar, tearoff=0)
filemenu.add_command(label="Exit", command=root.quit)
menubar.add_cascade(label="File", menu=filemenu)
root.config(menu=menubar)

pad_opts = {"padx": 10, "pady": 6}

# Row 1: Button and checkboxes
lbl_button = tk.Label(root, text="Button:", bg="#D4D0C8", anchor="w")
lbl_button.grid(row=0, column=0, sticky="w", **pad_opts)

btn_sample = tk.Button(
    root,
    text="Button",
    width=12,
    relief="raised",
    command=lambda: on_button_click("Button"),
)
btn_sample.grid(row=0, column=1, sticky="w", **pad_opts)

chk_frame = tk.Frame(root, bg="#D4D0C8")
chk_frame.grid(row=0, column=2, rowspan=3, sticky="nw", padx=30, pady=5)

chk_enabled = tk.Checkbutton(chk_frame, text="Enabled", bg="#D4D0C8")
chk_enabled.select()
chk_enabled.pack(anchor="w", pady=2)

chk_visible = tk.Checkbutton(chk_frame, text="Visible", bg="#D4D0C8")
chk_visible.select()
chk_visible.pack(anchor="w", pady=2)

chk_editable = tk.Checkbutton(chk_frame, text="Editable", bg="#D4D0C8")
chk_editable.select()
chk_editable.pack(anchor="w", pady=2)

# Row 2: Text field
lbl_textfield = tk.Label(root, text="TextField:", bg="#D4D0C8", anchor="w")
lbl_textfield.grid(row=1, column=0, sticky="w", **pad_opts)

entry_text = tk.Entry(root, width=30, relief="sunken")
entry_text.grid(row=1, column=1, sticky="w", **pad_opts)


def on_enter(event=None):
    messagebox.showinfo("입력값", f"입력된 값: {entry_text.get()}")


entry_text.bind("<Return>", on_enter)

# Entry event-binding examples:
# def on_click(event):
#     print("Click position:", event.x, event.y)
#     print("Screen position:", event.x_root, event.y_root)
#     print("Current value:", event.widget.get())
# entry_text.bind("<Button-1>", on_click)
# entry_text.bind("<ButtonRelease-1>", on_release)
# entry_text.bind("<Double-Button-1>", on_double)
# entry_text.bind("<Button-3>", on_right)
# entry_text.bind("<B1-Motion>", on_drag)
# entry_text.bind("<MouseWheel>", on_wheel)
# entry_text.bind("<Enter>", on_mouse_in)
# entry_text.bind("<Leave>", on_mouse_out)
# entry_text.bind("<FocusIn>", on_focus_in)
# entry_text.bind("<FocusOut>", on_focus_out)

# Row 3: Password field
lbl_password = tk.Label(root, text="PasswordField:", bg="#D4D0C8", anchor="w")
lbl_password.grid(row=2, column=0, sticky="w", **pad_opts)

entry_pass = tk.Entry(root, width=30, show="*", relief="sunken")
entry_pass.grid(row=2, column=1, sticky="w", **pad_opts)

# Row 4: Text area and property buttons
lbl_textarea = tk.Label(root, text="TextArea:", bg="#D4D0C8", anchor="w")
lbl_textarea.grid(row=3, column=0, sticky="nw", **pad_opts)

txt_frame = tk.Frame(root)
txt_frame.grid(row=3, column=1, sticky="w", **pad_opts)

scrollbar = tk.Scrollbar(txt_frame)
scrollbar.pack(side="right", fill="y")

text_area = tk.Text(
    txt_frame,
    width=28,
    height=7,
    relief="sunken",
    yscrollcommand=scrollbar.set,
)
text_area.insert(
    "1.0",
    "Use the checkboxes and\n"
    "buttons on the right to set the\n"
    "attributes of the widgets on the\n"
    "left.\n\n"
    "The font support is not hooked\n"
    "up yet.",
)
text_area.pack(side="left", fill="both")
scrollbar.config(command=text_area.yview)

btn_frame = tk.Frame(root, bg="#D4D0C8")
btn_frame.grid(row=3, column=2, sticky="nw", padx=30, pady=5)

buttons = [
    "BackgroundColor",
    "ForegroundColor",
    "Font",
    "ToolTip",
    "Background BackgroundColor",
]
for btn_text in buttons:
    # Capture this iteration's text in the callback's default argument.
    button = tk.Button(
        btn_frame,
        text=btn_text,
        width=22,
        relief="raised",
        command=lambda name=btn_text: on_button_click(name),
    )
    button.pack(pady=3, anchor="w")

# Row 5: Static text
lbl_static_lbl = tk.Label(root, text="StaticText:", bg="#D4D0C8", anchor="w")
lbl_static_lbl.grid(row=4, column=0, sticky="w", **pad_opts)

lbl_static_val = tk.Label(root, text="StaticText", bg="#D4D0C8")
lbl_static_val.grid(row=4, column=1, sticky="w", **pad_opts)

# Row 6: Checkbox
lbl_checkbox_lbl = tk.Label(root, text="CheckBox:", bg="#D4D0C8", anchor="w")
lbl_checkbox_lbl.grid(row=5, column=0, sticky="w", **pad_opts)

chk_sample = tk.Checkbutton(root, text="CheckBox", bg="#D4D0C8")
chk_sample.grid(row=5, column=1, sticky="w", **pad_opts)

# Row 7: Radio group and image area
lbl_radiogroup = tk.Label(root, text="RadioGroup:", bg="#D4D0C8", anchor="w")
lbl_radiogroup.grid(row=6, column=0, sticky="nw", **pad_opts)

radio_frame = tk.LabelFrame(root, text="A RadioBox", bg="#D4D0C8", padx=10, pady=5)
radio_frame.grid(row=6, column=1, sticky="w", **pad_opts)

radio_var = tk.StringVar(value="one")
r1 = tk.Radiobutton(radio_frame, text="one", value="one", variable=radio_var, bg="#D4D0C8")
r2 = tk.Radiobutton(radio_frame, text="two", value="two", variable=radio_var, bg="#D4D0C8")
r3 = tk.Radiobutton(radio_frame, text="three", value="three", variable=radio_var, bg="#D4D0C8")
r1.pack(anchor="w")
r2.pack(anchor="w")
r3.pack(anchor="w")

lbl_image = tk.Label(root, text="Image:", bg="#D4D0C8")
lbl_image.grid(row=6, column=2, sticky="nw", padx=30, pady=5)

canvas_img = tk.Canvas(
    root,
    width=100,
    height=100,
    bg="#A0B0C0",
    highlightthickness=1,
    highlightbackground="gray",
)
canvas_img.grid(row=6, column=2, sticky="nw", padx=90, pady=5)

for i in range(0, 100, 10):
    canvas_img.create_line(i, 0, i, 100, fill="#B0C0D0")
    canvas_img.create_line(0, i, 100, i, fill="#B0C0D0")

# Row 8: Choice and image button
lbl_choice = tk.Label(root, text="Choice:", bg="#D4D0C8", anchor="w")
lbl_choice.grid(row=7, column=0, sticky="w", **pad_opts)

combo_choice = ttk.Combobox(
    root,
    values=["one", "two", "three"],
    width=8,
    state="readonly",
)
combo_choice.set("two")
combo_choice.grid(row=7, column=1, sticky="w", **pad_opts)

lbl_imgbtn = tk.Label(root, text="ImageButton:", bg="#D4D0C8")
lbl_imgbtn.grid(row=7, column=2, sticky="w", padx=30, pady=5)

edit_icon = tk.PhotoImage(file="edit_icon.png")
btn_edit = tk.Button(
    root,
    text=" edit",
    image=edit_icon,
    compound="left",
    width=80,
    bg="#6A6A9B",
    fg="white",
    relief="raised",
    command=lambda: on_button_click("edit"),
)
btn_edit.image = edit_icon
btn_edit.grid(row=7, column=2, sticky="w", padx=120, pady=5)

# Row 9: List and borderless text field
lbl_list = tk.Label(root, text="List:", bg="#D4D0C8", anchor="w")
lbl_list.grid(row=8, column=0, sticky="nw", **pad_opts)

listbox = tk.Listbox(root, width=10, height=4, relief="sunken")
for item in ["one", "two", "three"]:
    listbox.insert("end", item)
listbox.select_set(2)
listbox.grid(row=8, column=1, sticky="w", **pad_opts)

lbl_noborder = tk.Label(root, text="TextField with no border", bg="#D4D0C8")
lbl_noborder.grid(row=8, column=2, sticky="nw", padx=30, pady=5)

# Row 10: Slider
lbl_slider = tk.Label(root, text="Slider:", bg="#D4D0C8", anchor="w")
lbl_slider.grid(row=9, column=0, sticky="w", **pad_opts)

slider = tk.Scale(
    root,
    from_=0,
    to=100,
    orient="horizontal",
    showvalue=False,
    width=10,
    length=200,
    bg="#D4D0C8",
    bd=1,
)
slider.grid(row=9, column=1, sticky="w", **pad_opts)

root.mainloop()
```
