# 11장 내장함수, 람다식, 제너레이터, 모듈

```python
# abs() 함수: 숫자의 절댓값 반환
i = -20
print(abs(i))

# all(): 시퀀스의 모든 항목이 참이면 True 반환(0은 거짓)
mylist = [1, 3, 4, 0]
print(all(mylist))

# any(): 시퀀스 객체에 있는 한 항목이라도 참인 경우 True 반환
print(any(mylist))

# bin(): 정수의 이진 표현을 반환하는 데 사용
y = bin(64)
print(y)

# eval(): 전달된 수식을 구문 분석하고 프로그램 내에서 수식의 값을 계산한다.
exp = input("파이썬 수식 입력: ")
print(eval(exp))

# sum(): 리스트에 존재하는 항목을 전부 더하여 합계 반환
print(sum([1, 2, 3]))

# map(): 반복 가능한 객체(리스트, 튜플 등)의 각 항목에 주어진 함수를 적용한 후 적용 결과 반환.
def square(n):
    return n*n

list1 = [1,2,3,4,5]
result = list(map(square, mylist))
print(result)
```

#### 정렬과 탐색

- sort() 메소드를 가지고 이 메소드는 리스트를 정렬된 상태로 변경
- sorted() 내장 함수는 반복 가능 객체로부터 정렬된 리스트 새롭게 생성함.
    - sorted(시퀀스, key함수, reverse여부): key 함수는 각 원소 크기 반환, reverse=True면 내림차순

```python
students = [
    ('홍길동', 3.9, 20260101),
    ('김철수', 3.0, 20261421),
    ('최자영', 4.3, 20262342),
]
print(sorted(students, key=lambda student: student[2]))
```

#### 이터레이터
- 이터러블 객체: for 루프와 함께 사용할 수 있는 여러 종류의 객체
- \_\_iter\_\_()은 이터러블 객체 자신을 반환한다.
- \_\_next\_\_()은 다음 반복을 위한 값을 반환한다. 더 이상 값 없으면 Stoplteration 예외 발생하면 된다.

#### 제너레이터
- 키워드  yield를 사용해 함수로부터 이터레이터를 생성하는 하나의 방법

```python
def myGenerator():
    yield 'first'
    yield 'second'
    yield 'third'

for word in myGenerator():
    print(word)
```
