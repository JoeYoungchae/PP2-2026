#### 이터레이터
- 이터러블 객체: for 루프와 함께 사용할 수 있는 여러 종류의 객체
- \_\_iter\_\_()은 이터러블 객체 자신을 반환한다.
- \_\_next\_\_()은 다음 반복을 위한 값을 반환한다. 더 이상 값 없으면 Stoplteration 예외 발생하면 된다.

```python
class MyCounter(object):
    # 생성자 메소드를 정의한다.
    def __init__(self, low, high):
        self.current = low
        self.high = high

    # 이터레이터 객체로서 자신을 반환한다.
    def __iter__(self):
        return self

    def __len__(self):
        return high-low+1

    def __next__(self):
        # current가 high보다 크면 StopIteration 예외를 발생한다.
        # current가 high보다 작으면 다음 값을 반환한다.
        if self.current > self.high:
            raise StopIteration
        else:
            self.current += 1
            return self.current - 1
# 위에서 정의된 MyCounter 객체를 for 루프에서 사용해보자.
c = MyCounter(1, 10)
for i in c:
    print(i, end=' ')
```

#### 제너레이터
- 키워드  yield를 사용해 함수로부터 이터레이터를 생성하는 하나의 방법
- 정의된 제너레이터를 for 루프에서 이터레이터를 사용할 수 있다.

---
이터레이터: 클래스를 이용해 반복 가능한 객체를 생성
제너레이터: 함수를 이용해 반복 가능한 객체를 생성

```python
def myGenerator():
    yield 'first'
    yield 'second'
    yield 'third'

for word in myGenerator():
    print(word)
```

```python
def MyCounterGen(low, high):
    while low <= high:
        yield low
        low += 1

for i in MyCounterGen(1, 10):
    print(i, end=' ')
```

```python
# 피보나치 이터레이터

class FibIterator:
    def __init__(self, a=1, b=0, maxValue=50):
        self.a = a
        self.b = b
        self.maxValue = maxValue

    def __iter__(self):
        return self

    def __next__(self):
        n = self.a + self.b
        if n > self.maxValue:
            raise StopIteration()
        self.a = self.b
        self.b = n
        return n

for i in FibIterator():
    print(i, end=' ')
```

#### 연산자 오버로딩
- 연산자를 메소드로 정의하는 것
    - +, == 연산자는 실제로 str 클래스 안에 정의된 메소드이다.

연산자|내부적인 함수 호출
---|---
덧셈|x.\_\_add__(y)
뺄셈|x.\_\_sub__(y)
곱셈|x.\_\_mul__(y)
지수|x.\_\_pow__(y)
나눗셈(실수)|x.\_\_truediv__(y)
나눗셈(정수)|x.\_\_floordiv__(y)
나머지|x.\_\_mod__(y)
작음|x.\_\_lt__(y)
작거나 같음|x.\_\_le__(y)
같음|x.\_\_eq__(y)
같지 않음|x.\_\_ne__(y)
큼|x.\_\_gt__(y)
크거나 같음|x.\_\_ge__(y)

```python
class Point:
    def __init__(self, x=0, y=0):
        self.x = x
        self.y = y

    def __add__(self, other):
        x = self.x + other.x
        y = self.y + other.y
        return Point(x, y)
    
    def __str__(self):
        return 'Point('+str(self.x)+','+str(self.y)+')'

p1 = Point(1, 2)
p2 = Point(3, 4)
print(p1+p2)
```

#### 모듈
- 함수나 변수 또는 클래스들을 모아 놓은 파일.
- 모듈 안에 있는 함수들은 import 문장으로 다른 모듈로 포함될 수 있다.
- 모듈 중 main 모듈은 최상위 수준에서 실행되는 스크립트를 의미한다.

```python
# 피보나치 수열 모듈

def fib(n):
    a, b = 0, 1
    while b < n:
        print(b, end=' ')
        a, b = b, a+b
    print()

def fib2(n):
    result = []
    a, b = 0, 1
    while b < n:
        result.append(b)
        a, b = b, a+b
    return result
```

fib()와 fib2() 함수가 있는 파일 fibo.py를 만들고 스파이더의 현재 작업 디렉토리에 저장하자.
파일 확장자는 반드시 .py여야한다. 이 파일이 바로 모듈이다.
모듈 안에서 모듈 이름은 __name__ 변수로 알 수 있다.

- fibo.py 모듈에 들어 있는 모든 정의는 다른 모듈로 import될 수 있다.
    - 'import fibo'를 입력하여 실행한다.

- 만약 함수를 사용할 때마다 모듈의 이름을 적어주는 것이 귀찮다면 "from 모듈 import 함수" 문장을 사용해도 된다.
<br>(모듈 이름없이 해당 모듈의 함수를 사용할 수 있다.)

- 'import fibo import *': 모듈 안의 모든 것을 의미한다.

- 모듈에 별칭을 만들어 사용할 수도 있다.
    - 'import mymodule as lib'

- 명령어 프롬프트 창에서 파이썬 모듈을 실행하려면 다음과 같이 한다.
