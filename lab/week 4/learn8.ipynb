#### 클래스란?
- 객체의 형태를 정의하는 틀과 같은 것. -> 하나의 클래스로 여러 객체 생성 가능
- 클래스 안에 정의된 함수를 메소드라고 한다. -> 객체의 동작을 나타냄.
    - 파이썬에서는 메소드 안에서 변수 이름 앞에 self.을 붙이고 값을 저장하면 인스턴스 변수가 된다.
- 클래스 안에 정의된 변수를 인스턴스 변수라고 한다. -> 객체의 속성을 나타냄.
- 인스턴스 변수와 메소드를 합쳐서 클래스의 멤버라고 한다.




##### 생성자 메소드
- 생성자는 객체를 초기화하는 메소드로서 클래스로 객체를 생성할 때 디폴트로 호출한다.
- 인스턴스 변수는 생성자 안에서 초기화해야 한다.
- 생성자의 이름은 항상 __init__으로 한다.


##### 인스턴스 변수와 지역 변수의 구별

- 메소드 안에서 self를 붙이지 않고 변수를 생성하면 지역 변수가 된다.<br>
클래스 안에서 다른 메소드를 호출하려면 항상 앞에 .self를 붙여야 한다.

```python
class Counter:      # 클래스 이니셜은 항상 대문자
    def __init__(self, initial_count=0): # 생성함수
        self.count = initial_count

    def increment(self, amount=1):
        self.count += amount

    def get_count(self):
        return self.count

def test_counter():
    myc = Counter()
    print(myc.get_count())
    myc.increment()
    print(myc.get_count())
    myc.increment()
    print(myc.get_count())
    for _ in range(10):
        myc.increment()
    print(myc.get_count())

    yourc = Counter(5)
    print(yourc.get_count())
    yourc.increment(4)
    print(yourc.get_count())

if __name__ == "__main__":
    test_counter()
```

# TV 클래스 추가
- 데이터: 현재 채널번호, 볼륨량, ON/OFF 여부
- 동작: 켜기, 끄기, 채널변경, 볼륨변경, 현재 TV 객체 내용 출력

- 클래스명: MyTelevision

```python
class MyTelevision:
    serialNumber = 0        # 이것이 클래스 변수.

    def __init__(self, channel=1, volume=10, is_on=False):
        self.channel = channel
        self.volume = volume
        self.is_on = is_on
        MyTelevision.serialNumber += 1      # 클래스 변수를 하나 증가.
        # 클래스 변수의 값을 객체의 시리얼 번호로 한다.
        self.myserial = MyTelevision.serialNumber

    def turn_on(self):
        self.is_on = True

    def turn_off(self):
        self.is_on = False

    def change_channel(self, channel):
        self.channel = channel

    def change_volume(self, volume):
        self.volume = volume

    def display_status(self):
        print(f"channel: {self.channel}, volume: {self.volume}, is on: {self.is_on}")

def test_television():
    mytv = MyTelevision(channel = 11)
    mytv.turn_on()
    mytv.display_status()
    mytv.turn_off()
    mytv.display_status()

if __name__ == "__main__":
    test_television()
```

* 클래스명: Person
* 속성: 이름(name:str), 출생년도(birth_year:int)
* 연산:
 - 네 이름은?   (get_name():str)
 - 올해 네 나이는?  (get_age():int)

테스트 함수 : test_person()
- 우리 팀의 4명에 대한 객체를 생성하고 목록에 저장: list 사용
- for 문을 통해 목록에 있는 모든 객체의 이름과 나이 출력

메인 함수
- 테스트 함수 수행

```python
from datetime import datetime

class Person:
    def __init__(self, name, birth_year):
        self.__name = name
        self.__birth_year = birth_year

    def get_name(self):
        return self.__name

    def get_age(self):
        current_year = self.__get_current_year()
        return current_year - self.__birth_year

    def __get_current_year(self):
        return datetime.now().year

    def __str__(self):
        return f"Name: {self.__name}, age: {self.get_age()}"

def test_person():
    song = Person("Song", 2001)
    y = song.__get_current_year()
    print(y)

    team_members = [Person("Kim", 2000), Person("Lee", 1999)]
    for member in team_members:
        print(member)

if __name__ == "__main__":
    test_person()
```

#### \_\_str__ 메소드
- 객체를 출력하였을 때 객체 안의 정보들을 출력한다.
- \_\_str__()메소드는 일반적으로 객체의 데이터를 문자열로 만들어서 반환한다.

```python
class Counter:
    def __init__(self, count=0):
        self.count = count
    def increment(self):
        self.count += 1
    def __str__(self):      # __str()__ 메소드는 객체를 print()로 출력 시 자동 호출
        msg = "카운트값: " + str(self.count)
        return msg

a = Counter(100)
print(a)
```

```python
# LAB) 백터 객체에 특수 메소드 정의하기

class Vector2D:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Vector2D(self.x + other.x, self.y + other.y)

    def __sub__(self, other):
        return Vector2D(self.x - other.x, self.y - other.y)

    def __eq__(self, other):
        return self.x == other.x and self.y == other.y

    def __str__(self):
        return '(%g, %g)' % (self.x, self.y)

u = Vector2D(0, 1)
v = Vector2D(1, 0)
w = Vector2D(1, 1)
a = u + v
print( a)
```
