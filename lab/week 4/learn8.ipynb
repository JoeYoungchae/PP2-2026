= 클래스의 인스턴스 = 객체

#### 클래스란?
- 객체의 형태를 정의하는 틀과 같은 것. -> 하나의 클래스로 여러 객체 생성 가능
- 클래스 안에 정의된 함수를 메소드라고 한다. -> 객체의 동작을 나타냄.
    - 파이썬에서는 메소드 안에서 변수 이름 앞에 self.을 붙이고 값을 저장하면 인스턴스 변수가 된다.
- 클래스 안에 정의된 변수를 인스턴스 변수라고 한다. -> 객체의 속성을 나타냄.
- 인스턴스 변수와 메소드를 합쳐서 클래스의 멤버라고 한다.




##### 생성자 메소드
- 생성자는 객체를 초기화하는 메소드로서 클래스로 객체를 생성할 때 디폴트로 호출한다.
- 인스턴스 변수는 생성자 안에서 초기화해야 한다.
- 생성자의 이름은 항상 __init__으로 한다.


##### 인스턴스 변수와 지역 변수의 구별

- 메소드 안에서 self를 붙이지 않고 변수를 생성하면 지역 변수가 된다.<br>
클래스 안에서 다른 메소드를 호출하려면 항상 앞에 .self를 붙여야 한다.

- 데이터: 현재 숫자
- 동작: 증가하기, 현재 값 읽어오기

```python
'''
객체의 멤버 접근

객체의 메소드를 호출하려면 객체 이름에 점을 붙이고 메소드 이름을 적어주면 된다.
변수에 접근할 때도 객체 이름에 점을 붙이고 변수 이름을 적어주면 된다.
'''
a = Counter()
a.increment()
a.count = 100
```

```python
class Counter:      # 클래스 이니셜은 항상 대문자
    def __init__(self, initial_count=0): # 생성함수
        self.count = initial_count

    def increment(self, amount=1):
        self.count += amount

    def get_count(self):
        return self.count

def test_counter():
    myc = Counter()
    print(myc.get_count())
    myc.increment()
    print(myc.get_count())
    myc.increment()
    print(myc.get_count())
    for _ in range(10):
        myc.increment()
    print(myc.get_count())

    yourc = Counter(5)
    print(yourc.get_count())
    yourc.increment(4)
    print(yourc.get_count())

if __name__ == "__main__":
    test_counter()
```

# TV 클래스 추가
- 데이터: 현재 채널번호, 볼륨량, ON/OFF 여부
- 동작: 켜기, 끄기, 채널변경, 볼륨변경, 현재 TV 객체 내용 출력

- 클래스명: MyTelevision

```python
class MyTelevision:
    serialNumber = 0

    def __init__(self, channel=1, volume=10, is_on=False):
        self.channel = channel
        self.volume = volume
        self.is_on = is_on
        MyTelevision.serialNumber += 1
        self.myserial = MyTelevision.serialNumber

    def turn_on(self):
        self.is_on = True

    def turn_off(self):
        self.is_on = False

    def change_channel(self, channel):
        self.channel = channel

    def change_volume(self, volume):
        self.volume = volume

    def display_status(self):
        print(f"channel: {self.channel}, volume: {self.volume}, is on: {self.is_on}")

def test_television():
    mytv = MyTelevision(channel = 11)
    mytv.turn_on()
    mytv.display_status()
    mytv.turn_off()
    mytv.display_status()

if __name__ == "__main__":
    test_television()
```

# 정보 은닉
- 가시성 처리 방법
- 구현의 세부 사항을 클래스 안에 감추는 것.
- 변수 뿐 아니라 메소드에도 정보 은닉을 사용할 수 있다.
- private: 변수나 함수 이름을 __로 시작한다.
- 정보 은닉의 필요성?
    1. 인스턴스 변수의 값이 올바르지 않게 변경될 수 있다.
    2. 클래스를 유지 보수하는 것이 어려워진다.
<br>
- 접근자: 인스턴스 변수값을 반환함.
    - 이러한 메소드엔 대개 get이 메소드 이름 앞에 붙여진다.(예: getAge())
- 설정자: 인스턴스 변수값을 설정함.
    - 이러한 메소드엔 대개 set이 메소드 이름 앞에 붙여진다.(예: setAge())
<br>

#### 접근자와 설정자 사용 이유
1. 나중에 클래스 업그레이드 시 편리
2. 접근자에서 매개 변수를 통해 잘못된 값이 넘어오는 경우, 이를 사전에 차단 가능
3. 필요할 때마다 인스턴스 변수값을 계산하여 반환 가능
4. 접근자만을 제공하면 자동적으로 읽기만 가능한 인스턴스 변수를 만들 수 있음.

```python
class Student:
    def __init__(self, name, age):
        self.__name = name
        self.__age = age

    def display_info(self):
        print(f"Name: {self.__name}, Age: {self.__age}")

kim = Student("Kim", 21)
kim.display_info()
print(kim.__age)
```

* 클래스명: Person
* 속성: 이름(name:str), 출생년도(birth_year:int)
* 연산:
 - 네 이름은?   (get_name():str)
 - 올해 네 나이는?  (get_age():int)

테스트 함수 : test_person()
- 우리 팀의 4명에 대한 객체를 생성하고 목록에 저장: list 사용
- for 문을 통해 목록에 있는 모든 객체의 이름과 나이 출력

메인 함수
- 테스트 함수 수행

```python
from datetime import datetime

class Person:
    def __init__(self, name, birth_year):
        self.__name = name
        self.__birth_year = birth_year

    def get_name(self):
        return self.__name

    def get_age(self):
        current_year = self.__get_current_year()
        return current_year - self.__birth_year

    def __get_current_year(self):
        return datetime.now().year

    def __str__(self):
        return f"Name: {self.__name}, age: {self.get_age()}"

def test_person():
    song = Person("Song", 2001)
    y = song.__get_current_year()
    print(y)

    team_members = [Person("Kim", 2000), Person("Lee", 1999)]
    for member in team_members:
        print(member)

if __name__ == "__main__":
    test_person()
```

2개의 변수가 동일한 객체를 참조하고 있는지 검사하려면 is 또는 is not을 사용한다.

```python
t = MyTelevision(11, 10, True)
s = t

if s is t:
    print("2개의 변수는 동일한 객체를 참조함.")

if s is not t:
    print("2개의 변수는 다른 객체를 참조함.")
```
