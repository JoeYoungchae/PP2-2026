타입힌트는 함수의 매개변수와 변환값의 타입을 지정하는 목적으로 사용하며,
엄격한 타입지정으로 인한 프로그래머의 의도를 명확히 전달하기 위함이다.
안써도 되지만 파이썬 13부터는 쓰는 것을 권장하고 있다.

```python
def gread(msg: str) -> str:
    print("In greet", msg)
    return msg + "!"

print(gread("안녕"))
```

```python
# 매개 변수와 인수의 타입이 다르면 에러가 발생한다.

def subtract(var_a: int, var_b: int) -> int:
    return var_a - var_b

subtract(3, '5')
```
