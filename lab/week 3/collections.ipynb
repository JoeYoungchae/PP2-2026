##### 자료구조
<br>자료들을 저장하는 여러 가지 구조들

+ 시퀀스: 요소로 구성, 요소 간에는 순서가 있고, 번호가 붙여져 있다.
    + 문자열
    + 리스트
    + 튜플

```python
# 리스트를 튜플로

myList = [1,2,3,4]
myTuple = tuple(myList)
myTuple
```

```python
t = ('apple', 'banana', 'cherry')       # 튜플 패킹
(s1, s2, s3) = t                        # 튜플 언패킹
```

- enumerate() : 번호를 매긴다.
<br>all(), any()

```python
fruits = ["apple", "banana", "cherry"]
for index, value in enumerate(fruits):
    print(index, value)

el = list(enumerate(fruits))        # list of tuples (index, value)
print(el)
```

##### 딕셔너리

```python
# 항목 추가하기

capitals = {}
capitals["Korea"] = "Seoul"
capitals["USA"] = "Washington D.C."
capitals["Japan"] = "Tokyo"
print(capitals)

# 항목 삭제하기

city = capitals.pop("Japan")        # 주어진 키를 가진 항목이 없으면 KeyError 발생

if "USA" in capitals:
    capitals.pop("USA")              # 주어진 키를 가진 항목이 없으면 KeyError 발생
```
