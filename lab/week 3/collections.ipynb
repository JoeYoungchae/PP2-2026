- enumerate() : 번호를 매긴다.
<br>all(), any()

```python
fruits = ["apple", "banana", "cherry"]
for index, value in enumerate(fruits):
    print(index, value)

el = list(enumerate(fruits))        # list of tuples (index, value)
print(el)
```

##### 딕셔너리

- 값(value)과 관련된 키(key)도 같이 저장한다.
    - 딕셔너리_이름 = { 키1:값1, 키2:값2, ...}

```python
# 항목 추가하기

capitals = {}
capitals["Korea"] = "Seoul"
capitals["USA"] = "Washington D.C."
capitals["Japan"] = "Tokyo"
print(capitals)

# 항목 삭제하기

city = capitals.pop("Japan")        # 주어진 키를 가진 항목이 없으면 KeyError 발생

if "USA" in capitals:
    capitals.pop("USA")              # 주어진 키를 가진 항목이 없으면 KeyError 발생
```

##### 딕셔너리 메소드

연산|설명
---|---
d = dict()|공백 딕셔너리 생성
d[key] = value|딕셔너리에 키와 값을 저장
v = d[key]|key에 해당되는 값 반환
d.get(key, default)|주어진 키를 가지고 값을 찾는다.(없으면 default 반환)
d.values()|딕셔너리 안의 모든 값의 시퀀스 반환
d.keys()|딕셔너리 안의 모든 키의 시퀀스 반환
d.items()|딕셔너리 안의 모든 (키, 값) 반환

```python
# 연락처 프로그램

def main():
    address_book = {}
    while True:
        user = display_menu()
        if user == 1:
            name, number = get_contact()
            address_book[name] = number
        elif user == 2:
            name, number = get_contact()
            address_book.pop(name)
        elif user == 3:
            pass
        elif user == 4:
            for key in sorted(address_book):
                print(key, "의 전화번호:", address_book[key])
        else:
            break

def get_contact():
    name = input("이름: ")
    number = input("전화번호: ")
    return name, number

def display_menu():
    print("1. 연락처 추가")
    print("2. 연락처 삭제")
    print("3. 연락처 검색")
    print("4. 연락처 출력")
    print("5. 종료")
    select = int(input("메뉴 항목 선택: "))
    return select

main()
```
